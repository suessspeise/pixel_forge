# Deepfryer

Toolkit for [a deepfried aesthetic](https://www.reddit.com/r/DeepFriedMemes).

```python
import sys, os
sys.path.insert(0, os.path.abspath("../src"))
from deepfry import Deepfryer

[func for func in dir(Deepfryer) if not func.startswith('__')]
```

```python
from PIL import Image
dirname = '../img/photo/'
files = [os.path.join(dirname,f) for f in os.listdir(dirname) if f.endswith('jpg')]
img = Image.open(files[5])
img.thumbnail((300,300))
img
```

```python
Deepfryer.add_noise(img)
```

```python
Deepfryer.blur(img)
```

```python
Deepfryer.blur(img, 'minfilter')
```

```python
Deepfryer.posterize(img)
```

```python
Deepfryer.reduce_color_space(img)
```

```python
Deepfryer.saturation_contrast(img)
```

```python
Deepfryer.tone_curving(img)
```

```python
Deepfryer.unsharp_masking(img)
```

```python
Deepfryer.jpeg_compression(img)
```

```python
Deepfryer.repeated_compression(img)
```

```python
Deepfryer.blend_overlay(img, Image.open(files[0]))
```

```python
Deepfryer.multilayer_overlay(img)
```

```python
temp = img.copy()
temp = Deepfryer.add_blob(temp, 100, 200, 100)
temp = Deepfryer.add_blob(temp, 100, 100, 40)
temp
```
